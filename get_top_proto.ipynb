```python
import torch
from PIL import Image
import os
import cv2
import numpy as np
import json
from matplotlib import pyplot as plt
import math
```

```python
# enter the directory to fMRI encoder model weights (for target brain area)
# and the directory storing representative images for prototypes 
proto_weight = torch.load('./log_mri2proto_slim_conv/VWFA-2/best_model.pt')['model']['proto_vector.weight'].cpu().numpy()
proto_dir = './log_prototype_coco_4_1_1000_128_slim/top_image_visualization_clipscore/'
```

```python
# select top-k prototypes for the targeted brain area for visualization
topk = 9
top_proto = np.argsort(proto_weight.mean(0))[::-1][:topk]

N = math.ceil(topk / 3)
fig, axes = plt.subplots(N, 3, figsize=(9, 3 * N), squeeze=False)

for ax, num in zip(axes.flat, top_proto):
    ax.imshow(Image.open(os.path.join(proto_dir, f"{num}.jpg")))
    ax.set_title(str(num))
    ax.axis("off")

for ax in axes.flat[topk:]:
    ax.axis("off")

plt.tight_layout()
```
