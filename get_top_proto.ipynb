```python
import torch
from PIL import Image
import os
import cv2
import numpy as np
import json
```

```python
# Configure local paths via project_paths.py or PROTOMRI_* environment variables.
from project_paths import NSD_ROOT, PROTOTYPE_RUN, FMRI_RUNS

subject = 'subj01'
roi_name = 'EBA'  # Example; change to your trained ROI (e.g., VWFA-2)
checkpoint = FMRI_RUNS / subject / roi_name / 'best_model.pt'
proto_weight = torch.load(checkpoint, map_location='cpu')['model']['proto_vector.weight'].cpu().numpy()
proto_dir = PROTOTYPE_RUN / 'top_image_visualization_clipscore' / 'raw'
```

```python
topk = 10
top_proto = np.argsort(proto_weight.mean(0))[::-1][:topk]
```

```python
# Display the image grids for the highest-weighted prototypes of this ROI.
# First run visualize_top_image.ipynb to generate these grids.
from IPython.display import display
for proto_id in top_proto:
    image_path = proto_dir / f'{int(proto_id)}.jpg'
    if image_path.exists():
        print(f'Prototype {proto_id}')
        display(Image.open(image_path))
    else:
        print(f'Missing grid: {image_path}')
```
